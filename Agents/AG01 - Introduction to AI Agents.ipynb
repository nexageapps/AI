## 1. What is an AI Agent?

### Definition

An **AI agent** is a system that:
1. **Perceives** its environment (receives input)
2. **Reasons** about what actions to take (makes decisions)
3. **Acts** on its environment (executes actions)
4. **Learns** from outcomes (adapts behavior)

### The Evolution: LLM Call → Chain → Agent

```
Simple LLM Call:
User: "What's 2+2?"
LLM: "2+2 equals 4."
✅ Fast, simple
❌ No tools, no planning, no verification

Chain:
User: "What's 2+2?"
Step 1: Parse question
Step 2: Generate SQL query
Step 3: Format result
✅ Multi-step processing
❌ Fixed sequence, no dynamic decisions

Agent:
User: "What's the weather in Paris and convert temp to Fahrenheit?"
Thought: I need weather data
Action: Call weather API for Paris
Observation: 20°C
Thought: Need to convert C to F
Action: Calculate (20 * 9/5) + 32
Observation: 68°F
Answer: "The weather in Paris is 20°C (68°F)"
✅ Dynamic planning, tool use, multi-step reasoning
❌ More complex, higher latency, higher cost
```

## 2. The ReAct Pattern (Reasoning + Acting)

The **ReAct** pattern is the foundation of most modern agents.

### ReAct Loop

```
1. Thought: "I need to find information about X"
2. Action: search("X")
3. Observation: [search results]
4. Thought: "Now I need to verify this information"
5. Action: calculator("verify equation")
6. Observation: [calculation result]
7. Thought: "I have enough information to answer"
8. Final Answer: [complete response]
```

### Example: Research Question

**Question:** "Who won the Nobel Prize in Physics in 2023 and how old were they?"

**ReAct Process:**

```
Thought 1: I need to search for 2023 Physics Nobel Prize winners
Action 1: search("2023 Nobel Prize Physics winners")
Observation 1: Pierre Agostini, Ferenc Krausz, Anne L'Huillier won for attosecond physics

Thought 2: I need to find their birth years to calculate age
Action 2: search("Pierre Agostini birth year")
Observation 2: Born 1941

Thought 3: Calculate age
Action 3: calculator("2023 - 1941")
Observation 3: 82

Thought 4: Repeat for other winners...
[similar process for Ferenc Krausz and Anne L'Huillier]

Thought 5: I have all information needed
Final Answer: "The 2023 Nobel Prize in Physics was awarded to Pierre Agostini (82), Ferenc Krausz (61), and Anne L'Huillier (65)..."
```

## 3. Agent Capabilities

### 3.1 Memory

Agents can remember:
- **Short-term:** Conversation history
- **Long-term:** Facts, preferences, past interactions
- **Episodic:** Specific events and their context
- **Semantic:** General knowledge extracted from experiences

**Example:**
```
User: "My name is Alice and I'm learning Python"
Agent: [stores: name=Alice, learning=Python]

[Later...]
User: "What was I learning?"
Agent: [retrieves memory] "You're learning Python"
```

### 3.2 Tools

Agents can use external tools:
- **Search:** Web search, database queries
- **Compute:** Calculators, code execution
- **APIs:** Weather, stocks, news, etc.
- **Actions:** Send emails, create files, update databases

**Example:**
```python
available_tools = [
    search_tool,      # Search the web
    calculator_tool,  # Perform calculations
    python_repl,      # Execute Python code
    database_tool,    # Query database
]
```

### 3.3 Planning

Agents can:
- **Decompose** complex tasks into subtasks
- **Sequence** actions in logical order
- **Adapt** plans based on intermediate results
- **Recover** from failures

**Example: Plan for "Create a sales report"**
```
1. Query database for sales data
2. Calculate summary statistics
3. Generate visualizations
4. Create PDF report
5. Email to stakeholders
```

### 3.4 Self-Correction

Agents can:
- **Verify** their outputs
- **Retry** failed actions
- **Ask for help** when uncertain
- **Learn** from mistakes

**Example:**
```
Action: calculator("10 / 0")
Observation: Error - Division by zero
Thought: My calculation failed, I need to handle this edge case
Action: Return "Cannot divide by zero" to user
```

## 4. When to Use Agents vs Simple Prompting

### Use Simple LLM Calls When:
- ✅ Task is single-step
- ✅ No external data needed
- ✅ Speed is critical
- ✅ Cost must be minimized
- ✅ Output format is simple

**Examples:**
- Text summarization
- Sentiment analysis
- Simple Q&A
- Text generation

### Use Chains When:
- ✅ Fixed multi-step sequence
- ✅ Each step is predictable
- ✅ No dynamic decision-making needed
- ✅ Moderate complexity

**Examples:**
- Document summarization pipeline
- Translation → Summarization
- Extract → Transform → Load

### Use Agents When:
- ✅ Task requires multiple steps
- ✅ Order of steps depends on intermediate results
- ✅ External tools needed
- ✅ Verification/self-correction required
- ✅ Task needs memory/context

**Examples:**
- Research and analysis
- Code debugging and fixing
- Data analysis with visualization
- Complex customer support
- Autonomous task completion

### Use Multi-Agent Systems When:
- ✅ Task requires specialized expertise
- ✅ Parallel processing beneficial
- ✅ Different perspectives needed
- ✅ Quality requires consensus/voting

**Examples:**
- Software development (designer + developer + tester)
- Content creation (researcher + writer + editor)
- Decision-making (multiple experts voting)

## 7. Limitations and Challenges

### Current Limitations

1. **Cost:** Agents use multiple LLM calls (expensive)
2. **Latency:** Multi-step reasoning takes time
3. **Reliability:** Can get stuck in loops or make wrong decisions
4. **Debugging:** Hard to trace why agent made certain choices
5. **Safety:** May take unintended actions

### Common Failure Modes

**Infinite Loops:**
```
Thought: I need more information
Action: search("X")
Observation: [results]
Thought: I need more information  # Same thought!
Action: search("X")  # Same action!
...(repeats forever)
```
**Solution:** Set max iterations, improve prompts

**Tool Misuse:**
```
User: "What's 2+2?"
Thought: I should search for this
Action: web_search("2+2")  # Wrong! Should use calculator
```
**Solution:** Better tool descriptions, examples

**Hallucination:**
```
Agent: "I found that X happened in 1950"
Reality: X happened in 1960
```
**Solution:** Verify facts, cite sources, use grounding

### Best Practices

1. **Set Limits:** Max iterations, timeouts, cost budgets
2. **Clear Instructions:** Detailed system prompts with examples
3. **Good Tools:** Well-described, reliable tools
4. **Monitoring:** Log all actions, set up alerts
5. **Human Oversight:** Human-in-the-loop for critical decisions
6. **Graceful Degradation:** Fallback to simpler methods if agent fails

## 8. Ethical Considerations

### Key Concerns

1. **Autonomy:** How much control should we give agents?
2. **Transparency:** Can users understand what the agent is doing?
3. **Accountability:** Who is responsible when agent makes mistakes?
4. **Privacy:** How does agent handle sensitive data?
5. **Bias:** Are agent decisions fair and unbiased?

### Guidelines

**DO:**
- ✅ Log all agent actions
- ✅ Allow human override
- ✅ Make limitations clear to users
- ✅ Implement safety checks
- ✅ Protect user data
- ✅ Test for bias

**DON'T:**
- ❌ Give agents unrestricted access
- ❌ Hide agent identity from users
- ❌ Allow irreversible actions without confirmation
- ❌ Store sensitive data unnecessarily
- ❌ Deploy without testing
- ❌ Ignore failure modes
